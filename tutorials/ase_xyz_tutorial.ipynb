# ASE Tutorial: Opening and Inspecting the xyz file

This notebook walks through:
1. Reading an extended XYZ file with ASE
2. Exploring basic `Atoms` properties
3. Understanding the `atoms.info` dictionary
4. Checking the correct shape of every data array
5. Iterating over all frames

The file is **XYZ** format — each frame's comment line contains JSON-serialised arrays that ASE automatically parses into `atoms.info`.

## 1. Imports

```python
import ase.io
import numpy as np
```

## 2. Reading the XYZ File

Pass `":"` as the index to read **all** frames, or `":10"` for the first 10.

```python
# Read all frames — returns a list of Atoms objects
db = ase.io.read("SINGLET_SOC_ALL.xyz", ":")

print(f"Total frames loaded : {len(db)}")

# Work with the first frame for exploration
atoms = db[0]
print(f"Type of each element: {type(atoms)}")
```

## 3. Basic Atoms Properties

```python
print("Number of atoms    :", len(atoms))
print("Chemical symbols   :", list(atoms.symbols))
print("Positions shape    :", atoms.positions.shape)   # (N_atoms, 3)
```

## 4. Listing All `atoms.info` Keys

`atoms.info` is a plain Python `dict`. Every key in the extXYZ comment line ends up here, with JSON arrays already parsed.

```python
print("Keys in atoms.info:")
for k in atoms.info:
    print(" ", k)
```

## 5. Shape Reference

The dataset contains **4 electronic states** and **N = 23 atoms**.  
The number of unique state pairs is `4 × (4-1) / 2 = 6`.

| Key | Expected shape | Meaning |
|---|---|---|
| `energy` | `(1, 4)` | 1 geometry × 4 state energies |
| `forces` | `(23, 4, 3)` | atoms × states × xyz |
| `couplings` | `(23, 6, 3)` | atoms × state pairs × xyz |
| `smooth_nacs` | `(23, 6, 3)` | smoothed NACs, same layout |
| `socs` | `(1, 240)` | flat SOC vector |

## 6. Inspecting Each Key

### 6.1 Energy — shape `(1, 4)`

```python
energy = np.array(atoms.info['REF_energy'])
print("energy shape :", energy.shape)   # (1, 4)
print("values       :", energy)
```

### 6.2 Forces — shape `(N_atoms, 4, 3)`

```python
forces = np.array(atoms.info['REF_forces'])
print("forces shape :", forces.shape)   # (23, 4, 3)

# Re-order to (n_states, N_atoms, 3) if you prefer state-first indexing
forces_sf = forces.transpose(1, 0, 2)
print("state-first  :", forces_sf.shape)
```

### 6.3 Couplings — shape `(N_atoms, 6, 3)`

```python
couplings = np.array(atoms.info['REF_smooth_nacs']) # Can adjust nacs key here
print("couplings shape :", couplings.shape)   # (23, 6, 3)
```

### 6.5 Spin–Orbit Couplings — shape `(1, 240)` with labels

```python
socs = np.array(atoms.info['REF_socs'])
print("socs shape   :", socs.shape)          # (1, 240)
```

## 7. Sanity-Check Script

Run this to verify all shapes are correct for every frame in the dataset.

```python
N        = len(atoms)                            # 23
n_states = 4
n_pairs  = n_states * (n_states - 1) // 2       # 6

expected_shapes = {
    'REF_energy'      : (1, n_states),
    'REF_forces'      : (N, n_states, 3),
    'REF_couplings'   : (N, n_pairs,  3),
    'REF_smooth_nacs' : (N, n_pairs,  3),
}

print(f"Checking frame 0 (N_atoms={N}, n_states={n_states}, n_pairs={n_pairs})\n")
all_ok = True
for key, expected in expected_shapes.items():
    actual = np.array(atoms.info[key]).shape
    ok = actual == expected
    status = "✓" if ok else f"✗  got {actual}"
    print(f"  {key:22s}  expected {str(expected):15s}  {status}")
    all_ok = all_ok and ok

print()
print("All shapes correct!" if all_ok else "Some shapes are wrong — check above.")
```

## 8. Iterating Over All Frames

```python
print(f"{'Frame':>6}  {'Min energy (eV)':>18}  {'Max |force| (eV/Å)':>20}")
print("-" * 50)

for i, frame in enumerate(db):
    e = np.array(frame.info['REF_energy'])
    f = np.array(frame.info['REF_forces'])
    print(f"{i:>6}  {e.min():>18.6f}  {np.abs(f).max():>20.6f}")
```
