# X-MACE Surface Hopping Workflow

This notebook documents how to set up and run an ensemble of SHARC surface hopping trajectories driven by an X-MACE model. This is **Step 2** of the overall workflow - it assumes `initconds.excited` already exists from Step 1 (the initial conditions tutorial).

## What this step does

```
initconds.excited          (from Step 1)
        │
        ▼
setup_traj.py              creates per-trajectory folders  →  traj-allmols/Singlet_1/TRAJ_*/
        │                                                      traj-allmols/Singlet_2/TRAJ_*/
        ▼
change_input-file.py       overwrites each SHARC input file with correct
        │                  nstates, initial state, charge, ezero, and seed
        ▼
resubmitting_and_changing_run-sh_AL.py
                           writes a run.sh and submits each trajectory to SLURM
```

All three steps are orchestrated by `submit-alltrajs_new.sh`, which is called once per structure folder.

## Directory layout expected

```
base_dir/
├── sh-scripts/
│   ├── MACE1.template          ← path to your trained model
│   ├── MACE.resources          ← (empty for CPU runs)
│   ├── input_setup_traj.txt    ← non-interactive input for setup_traj.py
│   ├── change_input-file.py    ← overwrites SHARC input files post-setup
│   └── resubmitting_and_changing_run-sh_AL.py  ← writes run.sh and submits
├── submit-alltrajs_new.sh
├── structure_0001/
│   └── initconds.excited
├── structure_0002/
│   └── initconds.excited
└── ...
```

**Reference:** SHARC Tutorial §2.7, González group, University of Vienna.  
https://sharc-md.org/wp-content/uploads/2021/04/SHARC_Tutorial.pdf

## Step 1 — Configure the MACE Interface Template

`MACE1.template` tells the SHARC MACE interface where to find your trained model. It contains a single line:

```
model_file /path/to/your/model.model
```

This is the only file you **must** edit before running the workflow. Update the path to point to your trained energy+forces model.

`MACE.resources` can be left empty for CPU runs. If you need GPU resources or memory limits, add them there.

````python
import os

os.makedirs("sh-scripts", exist_ok=True)

# ── MACE1.template — update the model path ───────────────────────────────
MODEL_PATH = "/path/to/your/energy_forces_model.model"   # <-- update this

with open("sh-scripts/MACE1.template", "w") as f:
    f.write(f"model_file {MODEL_PATH}\n")

# ── MACE.resources — leave empty for CPU runs ────────────────────────────
with open("sh-scripts/MACE.resources", "w") as f:
    f.write("")

print("sh-scripts/MACE1.template written")
print("sh-scripts/MACE.resources written (empty)")
````

## Step 2 — Configure `setup_traj.py` Inputs

`setup_traj.py` is an interactive SHARC script that reads `initconds.excited` and creates the per-trajectory folder structure. We run it non-interactively by piping `input_setup_traj.txt` to stdin.

### What the input file answers

The two values you need to change for your system are the **number of trajectories** and the **random seed**. Everything else accepts the default for a standard MACE-driven singlet dynamics run.

| Line | Value | Prompt answered |
|---|---|---|
| 1 | *(blank)* | Accept default `initconds.excited` filename |
| 2 | `2` | Number of trajectories per initial condition group |
| 3 | `-1` | Use all available initial conditions |
| 4–8 | *(blank)* | Accept defaults for timestep, simulation length, substeps, etc. |
| 9 | `1234` | Random seed for trajectory selection |
| 10 | `14` | Number of parallel tasks (match your SLURM `--cpus-per-task`) |
| 11–27 | *(blank)* | Accept all further defaults |
| 28 | `no` | Do not use adaptive time steps |
| 29–33 | *(blank)* | Accept defaults |
| 34 | `MACE1.template` | Path to the QC interface template file |
| 35 | `yes` | Confirm the template |
| 36 | `MACE.resources` | Path to the resources file |
| 37 | *(blank)* | Accept default |
| 38 | `no` | Do not copy large files into each trajectory folder |
| 39–52 | *(blank)* | Accept all remaining defaults |

````python
# Write input_setup_traj.txt
# The only values you should change are N_TRAJ and SEED.

N_TRAJ   = 2      # number of trajectories per initial condition group
N_ICONDS = -1     # -1 = use all available initial conditions
SEED     = 1234   # random seed for trajectory selection
N_TASKS  = 14     # parallel tasks — match your SLURM --cpus-per-task

lines = [
    "",              # accept default initconds.excited filename
    str(N_TRAJ),     # number of trajectories
    str(N_ICONDS),   # use all initial conditions
    "", "", "", "", "",
    str(SEED),       # random seed
    str(N_TASKS),    # parallel tasks
    "", "", "", "", "", "", "", "", "", "", "", "", "", "", "", "", "",
    "no",            # no adaptive time steps
    "", "", "", "", "",
    "MACE1.template", # interface template
    "yes",            # confirm
    "MACE.resources", # resources file
    "",
    "no",             # do not copy large files
    "", "", "", "", "", "", "", "", "", "", "", "", "", "",
]

with open("sh-scripts/input_setup_traj.txt", "w") as f:
    f.write("\n".join(lines) + "\n")

print("sh-scripts/input_setup_traj.txt written")
````

## Step 3 — Configure the SHARC Input Files

After `setup_traj.py` creates the trajectory folders, each `TRAJ_*/input` file contains a generic SHARC input generated from defaults. `change_input-file.py` overwrites every input file with the correct settings for your specific system.

### What it changes

The script walks the `traj-allmols/` directory tree, finds every folder named `Singlet_1` or `Singlet_2`, and for each `TRAJ_*/input` file it:

1. Reads the existing `rngseed` and `ezero` values (so they are preserved per-trajectory)
2. Sets `state` to `2 mch` (S1 excitation) for `Singlet_1` folders, or `3 mch` (S2 excitation) for `Singlet_2` folders
3. Rewrites the full input with a fixed set of dynamics parameters

### Key SHARC input parameters

| Parameter | Value | Meaning |
|---|---|---|
| `nstates 2 0 0` | 2 singlets | Number of electronic states (singlets, doublets, triplets) |
| `actstates 2 0 0` | 2 active | Active states included in propagation |
| `state 2 mch` | S1 | Initial state (1-indexed, MCH representation) |
| `tmax 7000` | 7 ps | Total simulation time in fs |
| `stepsize 0.5` | 0.5 fs | Nuclear timestep |
| `nsubsteps 25` | 25 | Electronic substeps per nuclear step |
| `surf diagonal` | — | Propagate in the diagonal (adiabatic) representation |
| `coupling ktdc` | — | Kinetic energy-based time-derivative coupling |
| `decoherence_scheme edc` | — | Energy-based decoherence correction |
| `hopping_procedure sharc` | — | SHARC fewest-switches surface hopping |
| `charge -1 0 0` | -1 for singlets | Molecular charge per multiplicity |
| `nospinorbit` | — | Ignore spin-orbit coupling (singlet-only dynamics) |

**To adapt for your system:** change `nstates`, `actstates`, `charge`, `tmax`, and the `state` assignments in the `Singlet_1`/`Singlet_2` logic. If you only have one initial state, you can simplify the `if/elif` block.

````python
# change_input-file.py
# Run from inside traj-allmols/ after setup_traj.py has created the TRAJ_* folders.
# Overwrites every TRAJ_*/input with the correct SHARC settings for this system.

import os
import re

top_level_dir = os.getcwd()   # run from inside traj-allmols/

# ── SHARC input template ──────────────────────────────────────────────────
# {state_line}, {rngseed}, and {ezero} are filled in per-trajectory below.
# Adjust nstates, actstates, tmax, charge, and coupling for your system.
input_template = """printlevel 2

geomfile "geom"
veloc external
velocfile "veloc"
nstates 2 0 0
actstates 2 0 0
{state_line}
coeff auto
rngseed {rngseed}
charge -1 0 0
ezero    {ezero}
tmax 7000.000000
stepsize 0.500000
nsubsteps 25
surf diagonal
coupling ktdc
gradcorrect
ekincorrect parallel_vel
reflect_frustrated none
decoherence_scheme edc
decoherence_param 0.1
hopping_procedure sharc
grad_all
nospinorbit
output_format ascii
output_dat_steps 1
"""

# ── State assignment per Singlet folder ───────────────────────────────────
# Singlet_1 = trajectories starting in S1 (state index 2 in 1-indexed MCH)
# Singlet_2 = trajectories starting in S2 (state index 3 in 1-indexed MCH)
STATE_MAP = {
    "Singlet_1": "state 2 mch",
    "Singlet_2": "state 3 mch",
}

updated = 0
for root, dirs, files in os.walk(top_level_dir):
    folder_name = os.path.basename(root)
    if folder_name not in STATE_MAP:
        continue
    state_line = STATE_MAP[folder_name]

    for subdir in dirs:
        input_path = os.path.join(root, subdir, "input")
        if not os.path.isfile(input_path):
            continue

        with open(input_path) as fh:
            content = fh.read()

        # Preserve the per-trajectory rngseed and ezero set by setup_traj.py
        m_seed  = re.search(r"rngseed\s+([-+]?\d+)", content)
        m_ezero = re.search(r"ezero\s+([-+]?\d*\.?\d+)", content)
        rngseed = m_seed.group(1)  if m_seed  else "24999"
        ezero   = m_ezero.group(1) if m_ezero else "0.0"

        new_content = input_template.format(
            state_line=state_line,
            rngseed=rngseed,
            ezero=ezero
        )
        with open(input_path, "w") as fh:
            fh.write(new_content)
        print(f"Updated: {input_path}")
        updated += 1

print(f"\n{updated} input files updated.")
````

## Step 4 — Write `run.sh` and Submit Trajectories

`resubmitting_and_changing_run-sh_AL.py` walks the `traj-allmols/` tree and for every `TRAJ_*/` folder that does **not** yet have an `output.lis` file (i.e. has not finished), it:

1. Writes a SLURM `run.sh` script
2. Immediately submits it with `sbatch`

The check for `output.lis` means the script is safe to re-run — it only submits trajectories that have not completed, making it useful for resubmitting failed or timed-out jobs.

Each trajectory runs:
```bash
$SHARC/driver.py -i mace input &> driver.log
```
where `-i mace` tells the SHARC driver to use the MACE interface and `input` is the SHARC input file updated in Step 3.

**Before running:** update `PARTITION`, `TIME`, and `MEM` below for your cluster.

````python
# resubmitting_and_changing_run-sh_AL.py
# Run from inside traj-allmols/ after change_input-file.py.
# Writes a SLURM run.sh for each incomplete trajectory and submits it.

import os
import subprocess

# ── Cluster settings — adjust for your system ────────────────────────────
PARTITION = "YOURPARTITION"   # <-- update
TIME      = "20:00:00"
MEM       = "14G"
# ─────────────────────────────────────────────────────────────────────────

root_directory = os.getcwd()   # run from inside traj-allmols/
submitted = 0

for folder in os.listdir(root_directory):
    folder_path = os.path.join(root_directory, folder)

    if folder not in ("Singlet_1", "Singlet_2"):
        continue

    for subfolder in os.listdir(folder_path):
        subfolder_path = os.path.join(folder_path, subfolder)
        if not os.path.isdir(subfolder_path):
            continue

        output_file = os.path.join(subfolder_path, "output.lis")
        if os.path.exists(output_file):
            continue   # already completed — skip

        slurm_script = f"""#!/bin/bash
#SBATCH --job-name={subfolder}-SH
#SBATCH --output={subfolder_path}/slurm_%j.out
#SBATCH --time={TIME}
#SBATCH --signal=TERM@300
#SBATCH -N 1
#SBATCH --cpus-per-task=1
#SBATCH --mem={MEM}
#SBATCH --partition={PARTITION}

cd {subfolder_path}

source ~/.bashrc

$SHARC/driver.py -i mace input &> driver.log
"""
        run_sh = os.path.join(subfolder_path, "run.sh")
        with open(run_sh, "w") as fh:
            fh.write(slurm_script)
        subprocess.run(["sbatch", run_sh])
        print(f"Submitted: {subfolder}")
        submitted += 1

print(f"\n{submitted} trajectories submitted.")
````

## Step 5 — Run the Full Workflow with `submit-alltrajs_new.sh`

Steps 2–4 above are orchestrated by a single shell script. For each structure folder you pass as an argument it:

1. Creates `traj-allmols/` inside the structure folder
2. Copies `initconds.excited` and all `sh-scripts/` files into it
3. Runs `setup_traj.py < input_setup_traj.txt`
4. Runs `change_input-file.py`
5. Runs `resubmitting_and_changing_run-sh_AL.py`
6. Reports how many jobs are pending / running in your queue

Run from `base_dir` as:
```bash
bash submit-alltrajs_new.sh structure_0001 structure_0002 structure_0003
```
or for all structures at once:
```bash
bash submit-alltrajs_new.sh structure_*/
```

**Prerequisites before running:**
- `$SHARC` environment variable must point to your SHARC installation (`source run_sharc.sh`)
- `sh-scripts/MACE1.template` must contain the correct model path (Step 1)
- `initconds.excited` must exist inside each structure folder (from Step 1 of the overall workflow)
- `PARTITION` in `resubmitting_and_changing_run-sh_AL.py` must match your cluster

````python
# submit-alltrajs_new.sh
# Orchestrates the full trajectory setup and submission for one or more structure folders.
# Run from base_dir as: bash submit-alltrajs_new.sh structure_0001 structure_0002 ...
#
# This cell shows the script contents for reference — execute it from the terminal, not here.

script = r"""#!/bin/bash

if [ $# -lt 1 ]; then
    echo "Usage: $0 <subdir1> <subdir2> ..."
    exit 1
fi

for SUBDIR in "$@"; do
    echo ">>> Entering $SUBDIR"
    if ! cd "$SUBDIR"; then
        echo "Directory $SUBDIR does not exist — skipping"
        continue
    fi

    # Create (or recreate) the trajectory directory
    rm -rf traj-allmols
    mkdir -p traj-allmols
    cd traj-allmols

    # Copy initconds and all sh-scripts into the working directory
    cp ../initconds.excited .
    cp ../../sh-scripts/* .

    # Step 2: create per-trajectory folders
    $SHARC/setup_traj.py < input_setup_traj.txt

    # Step 3: overwrite SHARC input files with correct settings
    python change_input-file.py

    # Step 4: write run.sh and submit each trajectory
    python resubmitting_and_changing_run-sh_AL.py

    cd ../..
    echo ">>> Done: $SUBDIR"
done

echo "pending: $(squeue --me | grep -c ' PD ')"  "running: $(squeue --me | grep -c ' R ')"
"""

with open("submit-alltrajs_new.sh", "w") as f:
    f.write(script)

import os
os.chmod("submit-alltrajs_new.sh", 0o755)
print("submit-alltrajs_new.sh written.")
print()
print("Run from base_dir as:")
print("  bash submit-alltrajs_new.sh structure_0001 structure_0002 ...")
print("or for all structures:")
print("  bash submit-alltrajs_new.sh structure_*/")
````

## What a Running Trajectory Looks Like

Once submitted, each `TRAJ_*/` folder contains:

| File | Contents |
|---|---|
| `input` | SHARC input (written by `change_input-file.py`) |
| `geom` | Initial geometry in SHARC format |
| `veloc` | Initial velocities in SHARC format |
| `run.sh` | SLURM submission script |
| `driver.log` | SHARC driver stdout (appears once running) |
| `output.lis` | Per-step energies, populations, hopping events (appears once running) |
| `output.xyz` | Trajectory geometry at each step |
| `output.dat` | Full binary output (energies, coefficients, gradients) |

A trajectory is considered **complete** when `output.lis` exists. The resubmission script in Step 4 uses this to avoid resubmitting finished trajectories.

### Checking job status

```bash
squeue --me                          # all your jobs
grep 'Step' structure_0001/traj-allmols/Singlet_1/TRAJ_00001/driver.log  # progress
```

### Resubmitting failed jobs

If jobs time out before completing, simply re-run `resubmitting_and_changing_run-sh_AL.py` from inside `traj-allmols/`. It will skip any folder that already has `output.lis` and resubmit only the incomplete ones:

```bash
cd structure_0001/traj-allmols
python resubmitting_and_changing_run-sh_AL.py
```
