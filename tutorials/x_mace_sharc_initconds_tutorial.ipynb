# X-MACE → SHARC: Generating Initial Conditions for Excited-State Dynamics

This notebook walks through generating a SHARC `initconds` file using X-MACE machine-learning models in place of expensive ab initio single-point calculations.

## How this fits into the standard SHARC workflow

In a conventional SHARC study the initial condition setup follows these steps (SHARC Tutorial §2.2–2.7, González group, University of Vienna):

| Step | Standard SHARC | This workflow |
|---|---|---|
| Geometry optimisation + frequencies | QC code (Molcas, ORCA, …) | same — used to generate the ground-state MD starting structure |
| Ground-state MD | `xTB` via ASE | `xTB` via ASE (Step 1 below) |
| Geometry + velocity sampling | `wigner.py` on a Molden file | snapshots from the MD trajectory (velocities already present) |
| Excited-state single points | `setup_init.py` + QC code | X-MACE energy model + X-MACE oscillator strength model (Steps 2–3) |
| Build `initconds` | output of `setup_init.py` | Step 4 below |
| Assign initial states | `excite.py` | `excite.py` — unchanged (Step 5) |
| Create trajectory folders | `setup_traj.py` | `setup_traj.py` — unchanged (Step 6) |

Steps 5 and 6 are standard SHARC and are documented in §2.5 and §2.7 of the SHARC Tutorial.

**Reference:** Mai, Marquetand & González, *WIREs Comput. Mol. Sci.* **8**, e1370 (2018).  
**SHARC Tutorial:** https://sharc-md.org/wp-content/uploads/2021/04/SHARC_Tutorial.pdf  
**SHARC GitHub:** https://github.com/sharc-md/sharc

## Step 1 — Ground-State MD with xTB

A ground-state NVE MD simulation is run with the GFN2-xTB semiempirical method via ASE. This serves the same purpose as Wigner sampling in the standard SHARC workflow — it generates an ensemble of geometries and velocities that sample the ground-state nuclear distribution.

**Why MD instead of Wigner sampling?**  
Wigner sampling from a harmonic frequency calculation works well for small rigid molecules but breaks down for large or flexible systems with anharmonic modes. Sampling directly from an MD trajectory is more appropriate for such systems (see SHARC Tutorial §3.4).

The script:
1. Reads a starting geometry from an XYZ file (passed as a command-line argument)
2. Relaxes it with BFGS until max force < 0.1 eV/Å
3. Initialises velocities from a Maxwell–Boltzmann distribution at 300 K, then removes net translation and rotation
4. Runs 100 000 NVE steps at 1 fs timestep, saving a snapshot every 10 steps → 10 000 frames in `md_trajectory.traj`

Run from the terminal as:
```bash
python xtb_md.py structure.xyz
```

````python
# xtb_md.py
# Run a ground-state GFN2-xTB geometry relaxation followed by NVE MD.
# Usage: python xtb_md.py <input.xyz>

import sys
from ase.io import read
from ase.optimize import BFGS
from ase.md.verlet import VelocityVerlet
from ase.md.velocitydistribution import MaxwellBoltzmannDistribution, Stationary, ZeroRotation
from ase.io.trajectory import Trajectory
from ase import units
from xtb.ase.calculator import XTB

if len(sys.argv) < 2:
    raise ValueError("Usage: python xtb_md.py <input.xyz>")

# ── Load structure ────────────────────────────────────────────────────────
atoms = read(sys.argv[1])
atoms.calc = XTB(method="GFN2-xTB")

# ── Geometry relaxation ───────────────────────────────────────────────────
print(f"Initial energy: {atoms.get_potential_energy():.4f} eV")
optimizer = BFGS(atoms, logfile="relaxation.log")
optimizer.run(fmax=0.1)   # converge when max force < 0.1 eV/Å
print(f"Relaxed energy: {atoms.get_potential_energy():.4f} eV")
atoms.write("relaxed_mol.xyz")

# ── Initialise velocities at 300 K ───────────────────────────────────────
MaxwellBoltzmannDistribution(atoms, temperature_K=300)
Stationary(atoms)     # remove net linear momentum
ZeroRotation(atoms)   # remove net angular momentum

# ── NVE MD ───────────────────────────────────────────────────────────────
dyn  = VelocityVerlet(atoms, 1 * units.fs)
traj = Trajectory("md_trajectory.traj", "w", atoms)

def print_status():
    print(f"Step {dyn.nsteps:5d}  E_pot={atoms.get_potential_energy():.4f} eV  "
          f"E_kin={atoms.get_kinetic_energy():.4f} eV")

dyn.attach(traj.write,    interval=10)    # save every 10 steps → 10 000 frames
dyn.attach(print_status, interval=1000)
dyn.run(100_000)
````

## Step 2 — Predict Multi-State Energies with X-MACE

The X-MACE energy model is run over the first 101 frames of the MD trajectory. The predicted energies (S0, S1, S2) are stored in `atoms.info['REF_energy']` with shape `(1, 3)` in eV and the trajectory is written to a new XYZ file.

**Important — `atoms.info` is fully replaced here:**
```python
atoms.info = {"REF_energy": energy_array}   # replaces everything
```
This is intentional — we discard any keys written by xTB and start clean. Velocities are safe because ASE stores them in `atoms.arrays['momenta']`, not `atoms.info`.

````python
from ase.io import read, write
import numpy as np
from mace.calculators import MACECalculator

ENERGY_MODEL = "/path/to/energy_model.model"   # <-- update
N_STATES     = 3                                # S0, S1, S2
N_FRAMES     = 101                              # frames to use (index 0:N_FRAMES)

# Convert .traj → .xyz so individual frames can be re-read
write("md_trajectory.xyz", read("md_trajectory.traj", index=":"))

atoms_list = read("md_trajectory.xyz", index=f"0:{N_FRAMES}")

calculator = MACECalculator(
    model_paths=ENERGY_MODEL,
    n_energies=N_STATES,
    device='cpu'
)

all_energies = []
for atoms in atoms_list:
    calculator.calculate(atoms)
    energy_array = np.array(calculator.results['energy']).reshape(1, -1)   # (1, N_STATES)
    all_energies.append(energy_array.flatten())
    # Replace atoms.info — velocities in atoms.arrays['momenta'] are unaffected
    atoms.info = {"REF_energy": energy_array}

np.savetxt("predicted_energies.txt", np.array(all_energies),
           header="Energies (eV): rows=geometries, columns=states (S0 S1 S2)")
write("md_traj_with-energies.xyz", atoms_list)
print(f"Written {len(atoms_list)} frames to 'md_traj_with-energies.xyz'")
print(f"REF_energy shape per frame: (1, {N_STATES})")
````

## Step 3 — Predict Oscillator Strengths with X-MACE

The X-MACE oscillator strength model adds `REF_osc-strength` (shape `(2,)`, for S0→S1 and S0→S2) to each frame's `atoms.info`. This uses key addition (`atoms.info[key] = ...`) rather than replacement, so `REF_energy` from Step 2 is preserved.

The oscillator strength model was trained with `--n_energies=2` — see the oscillator strength training tutorial for details on how this works.

````python
from ase.io import read, write
import numpy as np
from mace.calculators import MACECalculator

OSC_MODEL = "/path/to/oscillator_model.model"   # <-- update
N_OSC     = 2   # S0→S1, S0→S2 — must match --n_energies used during oscillator training

atoms_list = read("md_traj_with-energies.xyz", index=":")

calculator = MACECalculator(
    model_paths=OSC_MODEL,
    n_energies=N_OSC,
    device='cpu'
)

all_osc = []
for atoms in atoms_list:
    calculator.calculate(atoms)
    osc_array = np.array(calculator.results['energy']).reshape(-1)   # (N_OSC,)
    all_osc.append(osc_array)
    # Add to atoms.info — does NOT overwrite REF_energy from Step 2
    atoms.info["REF_osc-strength"] = osc_array

np.savetxt("predicted_osc-strengths.txt", np.array(all_osc),
           header="Oscillator strengths: rows=geometries, columns=transitions (S0->S1, S0->S2)")
write("md_traj_with-energies-and-osc.xyz", atoms_list)
print(f"Written {len(atoms_list)} frames to 'md_traj_with-energies-and-osc.xyz'")

# Verify both keys are present before proceeding
test = atoms_list[1]
for k, v in test.info.items():
    print(f"  atoms.info['{k}']: shape {np.array(v).shape}")
````

## Step 4 — Build the SHARC `initconds` File

Converts the XYZ file into the SHARC 4.0 initial conditions format.

**Frame 0 is the equilibrium geometry.** SHARC requires the file to open with an equilibrium geometry block with no `Index` or `States` header. Frames 1 onwards become the actual initial conditions. `Ninit` in the header is therefore `len(frames) - 1`.

**Unit conversions applied:**

| Quantity | From | To | Factor |
|---|---|---|---|
| Positions | Å | Bohr | `/ units.Bohr` |
| Velocities | ASE internal | Bohr / a.u. time | `× 0.00448998` |
| Energies | eV | Hartree | `× 0.036749322` |
| Kinetic energy | eV | Hartree | `/ units.Hartree` |

**`Eref`** in the header is the absolute S0 energy of frame 0 (the equilibrium geometry) in Hartree. **`Epot`** for each initial condition is `E_S0(frame_0) − E_S0(frame_i)` — relative to equilibrium, so the equilibrium frame has `Epot = 0`.

**States block** — one line per electronic state:
```
001    E_j_Ha  E_S0_Ha   0 0 0   0 0 0   dE_j_Ha   osc_j
```
The six zeros are NAC/SOC placeholders (not needed for initial conditions). S0 (`j=0`) always gets `osc = 0.0`; S1 gets `osc_strengths[0]` (S0→S1); S2 gets `osc_strengths[1]` (S0→S2).

````python
from ase.io import read
from ase.data import atomic_masses
from ase import units
import numpy as np

N_STATES    = 3       # number of electronic states to write (S0, S1, S2)
STATES_LINE = "5 0 0" # singlets doublets triplets declared in the file header
                       # must be consistent with what excite.py expects
EV_TO_HA    = 0.036749322
VEL_CONV    = 0.00448998   # ASE velocity units → Bohr / a.u. time

frames        = read("md_traj_with-energies-and-osc.xyz", index=":")
ref_energies0 = np.array(frames[0].info["REF_energy"]).flatten()   # eV, frame 0
Eref_ha       = ref_energies0[0] * EV_TO_HA                         # S0 of frame 0, Hartree

with open("initconds", "w") as f:

    # ── File header ───────────────────────────────────────────────────────
    f.write("SHARC Initial conditions file, version 4.0   <Excited>\n")
    f.write(f"Ninit     {len(frames) - 1}\n")   # frame 0 = equilibrium, not an IC
    f.write(f"Nato     {frames[0].get_number_of_atoms()}\n")
    f.write("Repr      MCH\n")
    f.write(f"Eref         {Eref_ha:.10f}\n")
    f.write("Eharm           0.0000000000\n")
    f.write(f"States    {STATES_LINE}\n")
    f.write("\n\nEquilibrium\n")

    for i, frame in enumerate(frames):
        ref_energies  = np.array(frame.info["REF_energy"]).flatten()        # (N_STATES,) eV
        osc_strengths = np.array(frame.info["REF_osc-strength"]).flatten()  # (N_OSC,)

        pos_bohr = frame.get_positions() / units.Bohr
        vel_au   = frame.get_velocities() * VEL_CONV
        ekin_ha  = frame.get_kinetic_energy() / units.Hartree
        epot_ha  = (ref_energies0[0] - ref_energies[0]) * EV_TO_HA
        etot_ha  = epot_ha + ekin_ha

        atom_lines = [
            f"{atom.symbol} {atom.number:.1f} "
            f"{pos[0]:.8f} {pos[1]:.8f} {pos[2]:.8f} "
            f"{atomic_masses[atom.number]:.8f} "
            f"{vel[0]:.8f} {vel[1]:.8f} {vel[2]:.8f}\n"
            for atom, pos, vel in zip(frame, pos_bohr, vel_au)
        ]

        if i == 0:
            # Equilibrium geometry block — no Index / States header
            f.writelines(atom_lines)
            f.write("\n\n")
        else:
            f.write(f"Index     {i}\n")
            f.write("Atoms\n")
            f.writelines(atom_lines)
            f.write("States\n")
            for j in range(N_STATES):
                e_ha     = float(ref_energies[j])                    * EV_TO_HA
                ref_e_ha = float(ref_energies[0])                    * EV_TO_HA
                delta_ha = float(ref_energies[j] - ref_energies[0]) * EV_TO_HA
                osc      = 0.0 if j == 0 else float(osc_strengths[j - 1])
                f.write(
                    f"{j+1:03}    {e_ha:.8f} {ref_e_ha:.8f}   "
                    f"0.00000000   0.00000000   0.00000000   "
                    f"0.00000000   0.00000000   0.00000000   "
                    f"{delta_ha:.8f}   {osc:.8f}\n"
                )
            f.write(f"Ekin    {ekin_ha:.8f} a.u.\n")
            f.write(f"Epot_harm    0.00000000 a.u.\n")
            f.write(f"Epot    {epot_ha:.8f} a.u.\n")
            f.write(f"Etot_harm    {ekin_ha:.8f} a.u.\n")
            f.write(f"Etot    {etot_ha:.8f} a.u.\n")
            f.write("\n\n")

print(f"Written {len(frames)-1} initial conditions + 1 equilibrium geometry to 'initconds'")
````

## Step 5 — Assign Initial Electronic States with `excite.py` (standard SHARC)

This is the standard SHARC `excite.py` (SHARC Tutorial §2.5). It reads `initconds`, uses the oscillator strengths in the States block to stochastically assign each initial condition to an electronic state, and writes `initconds.excited`.

### Running non-interactively

`excite.py` is an interactive script, but all prompts can be answered by piping a text file to stdin. This is how the batch submission works. Create a file called `excite_inp.txt` with the following contents — the only two lines you need to change for your system are the **energy window** and the **seed**:

```
                  ← Enter: accept default initconds filename
                  ← Enter: accept default gen_list=3 (keep existing excited-state info already in initconds)
                  ← Enter: accept default excite=3 (delta-pulse, weighted by oscillator strength)
1.6 3.3           ← excitation energy window in eV (lower upper) — CHANGE THIS for your system
                  ← Enter: accept default — include all states in selection
1234              ← random number seed — CHANGE THIS or use ! for system time
                  ← Enter: accept default — overwrite output file
```

**Excitation window:** set the lower and upper bounds in eV to match the absorption band you want to simulate. Only geometries whose excitation energy falls within this window will be assigned an excited state; the rest are discarded. For our system the window is **1.6–3.3 eV**.

**Seed:** any integer gives a reproducible selection. Use `!` to seed from system time if you want a different selection each run.

Run from the terminal:
```bash
python excite.py < excite_inp.txt
```

Output: `initconds.excited` — contains everything from `initconds` plus the assigned initial state for each trajectory.

> SHARC also saves all your answers to `KEYSTROKES.excite` so you can inspect or reproduce the run.

````python
# Write excite_inp.txt
# Edit EWIN_LOW, EWIN_HIGH, and SEED for your system.

EWIN_LOW  = 1.6    # lower bound of excitation window, eV
EWIN_HIGH = 3.3    # upper bound of excitation window, eV
SEED      = 1234   # random seed; use the string '!' to seed from system time

excite_inp = f"""\n\n\n{EWIN_LOW} {EWIN_HIGH}\n\n{SEED}\n\n"""

with open("excite_inp.txt", "w") as f:
    f.write(excite_inp)

print("excite_inp.txt written. Run excite.py with:")
print("  python excite.py < excite_inp.txt")
````

## Step 6 — Create Trajectory Folders with `setup_traj.py` (standard SHARC)

This is the standard SHARC `setup_traj.py` (SHARC Tutorial §2.7). It reads `initconds.excited` and a SHARC input template and creates one folder per trajectory (`TRAJ_00001/`, `TRAJ_00002/`, …), each self-contained with its own `input`, `geom`, `veloc`, and run script.

You will need:
- `initconds.excited` — from Step 5
- A `SHARC.inp` template — specifies timestep, simulation time, propagation method, states, etc.
- A QM run script — the shell script that calls your QC interface (or the X-MACE interface) at each dynamics step
- A QC template file — e.g. `MOLCAS.template` or equivalent for your interface

Run interactively from the terminal:
```bash
mkdir traj
cd traj
setup_traj.py
```

`setup_traj.py` will prompt for: the `initconds.excited` filename, number of trajectories, integrator (velocity Verlet), timestep, simulation length, and the path to the QM run script and template files.

Once complete, run a trajectory with:
```bash
cd TRAJ_00001
sharc.x input > output &
```

Or submit the generated run scripts to your cluster scheduler.

> `setup_traj.py` saves all answers to `KEYSTROKES.setup_traj` for reproducibility.

See SHARC Tutorial §2.7 and SHARC Manual §6.6 for the full `SHARC.inp` reference.

## Batch Submission — Running Steps 2–5 on an HPC Cluster

When generating initial conditions for many different structures (e.g. a screening campaign), Steps 2–5 can be automated with a SLURM submission script. The code below iterates over structure folders and submits one job per structure.

Each job:
1. Runs the energy prediction (Step 2)
2. Runs the oscillator strength prediction (Step 3)
3. Builds `initconds` (Step 4)
4. Runs `excite.py` non-interactively (Step 5)

`setup_traj.py` (Step 6) is left separate because it typically requires interactive choices about the SHARC interface that vary per project.

**Directory layout expected:**
```
base_dir/
├── excite.py
├── excite_inp.txt
├── energy_model.model
├── oscillator_model.model
├── structure_010_mol1/
│   └── md_trajectory.traj
├── structure_010_mol2/
│   └── md_trajectory.traj
└── ...
```

````python
import os
import subprocess

base_dir      = os.getcwd()
ENERGY_MODEL  = f"{base_dir}/energy_model.model"       # <-- update
OSC_MODEL     = f"{base_dir}/oscillator_model.model"   # <-- update
FOLDER_PREFIX = "structure_01"   # submit jobs for all folders starting with this

slurm_template = """#!/bin/bash
#SBATCH --job-name=initconds_{folder}
#SBATCH --time=01:00:00
#SBATCH --signal=TERM@300
#SBATCH -N 1
#SBATCH --cpus-per-task=1
#SBATCH --mem=4G
#SBATCH --partition=barnard

source ~/.bashrc
source /projects/p_brighten/run_sharc3.sh

# ── Step 2: predict multi-state energies ──────────────────────────────────
python3 - << 'PYEOF'
from ase.io import read, write
import numpy as np
from mace.calculators import MACECalculator

N_STATES = 3
N_FRAMES = 101
write("md_trajectory.xyz", read("md_trajectory.traj", index=":"))
atoms_list = read("md_trajectory.xyz", index=f"0:{{N_FRAMES}}")
calculator = MACECalculator(model_paths="{energy_model}", n_energies=N_STATES, device="cpu")
for atoms in atoms_list:
    calculator.calculate(atoms)
    energy_array = np.array(calculator.results["energy"]).reshape(1, -1)
    atoms.info = {{"REF_energy": energy_array}}
write("md_traj_with-energies.xyz", atoms_list)
PYEOF

# ── Step 3: predict oscillator strengths ──────────────────────────────────
python3 - << 'PYEOF'
from ase.io import read, write
import numpy as np
from mace.calculators import MACECalculator

N_OSC = 2
atoms_list = read("md_traj_with-energies.xyz", index=":")
calculator = MACECalculator(model_paths="{osc_model}", n_energies=N_OSC, device="cpu")
for atoms in atoms_list:
    calculator.calculate(atoms)
    atoms.info["REF_osc-strength"] = np.array(calculator.results["energy"]).reshape(-1)
write("md_traj_with-energies-and-osc.xyz", atoms_list)
PYEOF

# ── Step 4: build initconds ────────────────────────────────────────────────
python3 - << 'PYEOF'
from ase.io import read
from ase.data import atomic_masses
from ase import units
import numpy as np

N_STATES    = 3
STATES_LINE = "5 0 0"
EV_TO_HA    = 0.036749322
VEL_CONV    = 0.00448998

frames        = read("md_traj_with-energies-and-osc.xyz", index=":")
ref_energies0 = np.array(frames[0].info["REF_energy"]).flatten()
Eref_ha       = ref_energies0[0] * EV_TO_HA

with open("initconds", "w") as f:
    f.write("SHARC Initial conditions file, version 4.0   <Excited>\\n")
    f.write(f"Ninit     {{len(frames) - 1}}\\n")
    f.write(f"Nato     {{frames[0].get_number_of_atoms()}}\\n")
    f.write("Repr      MCH\\n")
    f.write(f"Eref         {{Eref_ha:.10f}}\\n")
    f.write("Eharm           0.0000000000\\n")
    f.write(f"States    {{STATES_LINE}}\\n")
    f.write("\\n\\nEquilibrium\\n")
    for i, frame in enumerate(frames):
        ref_e = np.array(frame.info["REF_energy"]).flatten()
        osc   = np.array(frame.info["REF_osc-strength"]).flatten()
        pos   = frame.get_positions() / units.Bohr
        vel   = frame.get_velocities() * VEL_CONV
        ekin  = frame.get_kinetic_energy() / units.Hartree
        epot  = (ref_energies0[0] - ref_e[0]) * EV_TO_HA
        atom_lines = [
            f"{{a.symbol}} {{a.number:.1f}} {{p[0]:.8f}} {{p[1]:.8f}} {{p[2]:.8f}} "
            f"{{atomic_masses[a.number]:.8f}} {{v[0]:.8f}} {{v[1]:.8f}} {{v[2]:.8f}}\\n"
            for a, p, v in zip(frame, pos, vel)
        ]
        if i == 0:
            f.writelines(atom_lines)
            f.write("\\n\\n")
        else:
            f.write(f"Index     {{i}}\\n")
            f.write("Atoms\\n")
            f.writelines(atom_lines)
            f.write("States\\n")
            for j in range(N_STATES):
                e_ha  = float(ref_e[j])             * EV_TO_HA
                r_ha  = float(ref_e[0])             * EV_TO_HA
                de_ha = float(ref_e[j] - ref_e[0])  * EV_TO_HA
                o     = 0.0 if j == 0 else float(osc[j-1])
                f.write(f"{{j+1:03}}    {{e_ha:.8f}} {{r_ha:.8f}}   "
                        f"0.00000000   0.00000000   0.00000000   "
                        f"0.00000000   0.00000000   0.00000000   "
                        f"{{de_ha:.8f}}   {{o:.8f}}\\n")
            f.write(f"Ekin    {{ekin:.8f}} a.u.\\n")
            f.write(f"Epot_harm    0.00000000 a.u.\\n")
            f.write(f"Epot    {{epot:.8f}} a.u.\\n")
            f.write(f"Etot_harm    {{ekin:.8f}} a.u.\\n")
            f.write(f"Etot    {{epot+ekin:.8f}} a.u.\\n")
            f.write("\\n\\n")
PYEOF

# ── Step 5: assign initial electronic states ───────────────────────────────
python3 ../excite.py < ../excite_inp.txt
"""

submitted = 0
for folder in sorted(os.listdir(base_dir)):
    if not folder.startswith(FOLDER_PREFIX) or not os.path.isdir(folder):
        continue
    folder_path = os.path.join(base_dir, folder)
    slurm_path  = os.path.join(folder_path, "initconds.slurm")
    with open(slurm_path, "w") as f:
        f.write(slurm_template.format(
            folder=folder,
            energy_model=ENERGY_MODEL,
            osc_model=OSC_MODEL
        ))
    os.chmod(slurm_path, 0o755)
    subprocess.run(["sbatch", "initconds.slurm"], cwd=folder_path)
    print(f"Submitted: {folder}")
    submitted += 1

print(f"\n{submitted} jobs submitted.")
````
