# X-MACE: Training and Predicting Oscillator Strengths

This notebook shows how to train an X-MACE model to predict **oscillator strengths** — and how to run predictions with the trained model using the ASE calculator interface.

### The core idea

Oscillator strengths are **scalar values per electronic state transition**, just like energies. This means we can exploit X-MACE's existing multi-state energy machinery directly, with no new model architecture required:

1. **Pack all oscillator strengths into the `energy` field** of the XYZ file, with the `REF_energy` field holding the corresponding reference values.
2. **Set `--n_energies`** to the total number of oscillator strength values per frame.
3. Train with `--energy_weight` — the model learns the oscillator strengths as a vector of scalars, exactly as it would learn a multi-state energy vector.
4. At inference time, `calc.results["energy"]` gives you back the predicted oscillator strengths.

Because oscillator strengths are geometry-dependent scalars, the energy readout head (which maps atom-centred features to a sum of scalars) is a natural fit — there is no need for forces, so `--forces_weight=0.0`.

## 1. Understanding the Oscillator Strength Data Layout

For a molecule with `n_states` electronic states, the number of distinct **state-to-state transitions** (and thus oscillator strengths) from the ground state is `n_states - 1`. If you have oscillator strengths for all unique pairs, the count is `n_states * (n_states - 1) / 2`.

In this tutorial we assume you have one oscillator strength per excited state (i.e. transitions from S0 → S1, S0 → S2, S0 → S3 for a 4-state system), giving **3 oscillator strength values per frame**. Adjust `N_OSC` below for your system.

The key mapping is:

| XYZ field | Content | Shape | When present |
|---|---|---|---|
| `REF_energy` | reference oscillator strengths | `(1, N_OSC)` | Always, in the dataset file |
| `calc.results['energy']` | predicted oscillator strengths | `(N_OSC,)` | At inference time only, from the calculator |

All other fields (`forces`, `smooth_nacs`, `socs`, …) are **not used** — set their loss weights to `0.0` when training.

```python
import ase.io
import numpy as np

# Number of oscillator strengths per frame.
# For S0→S1, S0→S2, S0→S3 in a 4-state (S0–S3) system: N_OSC = 3
N_OSC = 3

# Load a few frames to inspect
db    = ase.io.read("oscillator_strengths.xyz", ":5")
atoms = db[0]

print(f"Atoms per frame  : {len(atoms)}")
print(f"Oscillator values: {N_OSC}  (one per S0→Sn transition)")
print()

# Only REF_energy exists in the raw dataset — it holds the reference
# oscillator strengths. atoms.info['energy'] does NOT exist here;
# 'energy' is only populated at inference time via calc.results['energy'].
ref = np.array(atoms.info['REF_energy'])

print(f"REF_energy shape : {ref.shape}   (expected (1, {N_OSC}))")
print(f"Reference oscillator strengths (frame 0): {ref.flatten()}")
print()
print("Keys available in atoms.info:")
for k in atoms.info:
    print(f"  {k}")
```

## 2. Preparing the XYZ Dataset

If your raw data has oscillator strengths stored under a different key (e.g. `osc_strengths`), the cells below show how to re-write the XYZ file so that oscillator strengths appear in the `energy` / `REF_energy` fields that X-MACE reads by default.

> **If your data already uses `energy` / `REF_energy` for oscillator strengths, skip this section.**

```python
# --- Pack oscillator strengths into REF_energy ---
#
# X-MACE reads reference targets from 'REF_energy'. If your raw dataset
# stores oscillator strengths under a different key, this cell renames them.
# atoms.info['energy'] is intentionally left absent — X-MACE does not need
# it for training and the model will never write predictions to atoms.info.
#
# Adjust SOURCE_FILE and SOURCE_KEY to match your raw data.

SOURCE_FILE = "raw_data.xyz"                 # <-- your original file
OUTPUT_FILE = "oscillator_strengths.xyz"     # <-- file to write for training
SOURCE_KEY  = "osc_strengths"                # key holding the oscillator strengths

frames = ase.io.read(SOURCE_FILE, ":")

for frame in frames:
    osc_vals = np.array(frame.info[SOURCE_KEY])   # shape (N_OSC,) or (1, N_OSC)

    # Store as REF_energy with shape (1, N_OSC) — the layout X-MACE expects
    frame.info['REF_energy'] = osc_vals.reshape(1, -1)

    # Remove the old key
    frame.info.pop(SOURCE_KEY, None)

ase.io.write(OUTPUT_FILE, frames)
print(f"Written {len(frames)} frames to '{OUTPUT_FILE}'")
print(f"Each frame: REF_energy shape = (1, {N_OSC})")
```

### 2b. Sanity check — verify shapes across all frames

```python
frames = ase.io.read("oscillator_strengths.xyz", ":")
print(f"Total frames: {len(frames)}")
print()

# Only REF_energy should be present. We do NOT check for atoms.info['energy']
# because that key does not exist in the dataset - predictions are written
# to calc.results['energy'] at inference time, never to atoms.info.
all_ok = True
for i, frame in enumerate(frames):
    if 'REF_energy' not in frame.info:
        print(f"  Frame {i:4d}  missing REF_energy  ← ERROR")
        all_ok = False
        continue
    ref_shape = np.array(frame.info['REF_energy']).shape
    expected  = (1, N_OSC)
    ok = ref_shape == expected
    if not ok:
        print(f"  Frame {i:4d}  REF_energy={ref_shape}  expected={expected}  ← WRONG")
        all_ok = False

if all_ok:
    print(f"All {len(frames)} frames have REF_energy shape (1, {N_OSC}). Ready to train.")
else:
    print("Some frames have issues — fix them before training.")
```

## 3. Training the Model

The training command is almost identical to a standard energy-only X-MACE run, with three key changes:

| Change | Reason |
|---|---|
| `--n_energies=3` | tells the model there are 3 scalar outputs per frame (= N_OSC) |
| `--forces_weight=0.0` | oscillator strengths have no forces |
| `--energy_weight=100.0` | the main (and only) loss term |

Everything else — model architecture, optimiser, scheduler — is unchanged.

> **Why does this work?**  
> The energy readout head in X-MACE/MACE maps atom-centred equivariant features through an MLP to produce a per-atom scalar, which is then summed over all atoms to give a **global scalar per state**. Because oscillator strengths are also geometry-dependent global scalars, the same machinery learns them without modification. The only difference is physical interpretation — the model does not know or care that the numbers are oscillator strengths rather than energies.

### 3a. AutoencoderExcitedMACE

```python
xmace_osc_cmd = """
python scripts/run_train.py \\
  --name="oscillator_strengths" \\
  --train_file="oscillator_strengths.xyz" \\
  --seed=100 \\
  --valid_fraction=0.1 \\
  --E0s='average' \\
  --model="AutoencoderExcitedMACE" \\
  --r_max=5.0 \\
  --batch_size=10 \\
  --n_energies=3 \\
  --correlation=3 \\
  --max_num_epochs=100 \\
  --ema \\
  --lr=0.0001 \\
  --ema_decay=0.99 \\
  --default_dtype="float32" \\
  --device=cuda \\
  --hidden_irreps="128x0e + 128x1o" \\
  --MLP_irreps='128x0e' \\
  --num_radial_basis=8 \\
  --num_interactions=2 \\
  --energy_weight=100.0 \\
  --forces_weight=0.0 \\
  --error_table="EnergyNacsDipoleMAE"
"""
# N_OSC = 3  -->  --n_energies=3
# forces_weight=0.0 because oscillator strengths have no associated forces

print("X-MACE oscillator strength training command:")
print(xmace_osc_cmd)

# Uncomment to run from this notebook:
# import subprocess
# subprocess.run(xmace_osc_cmd, shell=True, check=True)
```

### 3c. Training curves

Use the helper below to plot the training loss once a run has completed.

```python
import os
import sys
import json
from collections import defaultdict
import matplotlib.pyplot as plt


def parse_loss_log(log_path):
    """Parse training (mode='opt') and validation (mode='eval') losses from an
    X-MACE results log.

    The log is JSON-lines: one JSON object per line. Relevant keys:
        loss   -- the (weighted) loss value
        mode   -- 'opt' for per-batch training steps, 'eval' for validation
        epoch  -- integer epoch, or null for the initial pre-training eval

    Returns (train_epochs, train_loss, valid_epochs, valid_loss) where the
    per-epoch training loss is the mean of that epoch's batch losses.
    """
    if not os.path.exists(log_path):
        print(f"Log file not found: {log_path}")
        return None

    opt_by_epoch = defaultdict(list)  # epoch -> [per-batch training losses]
    valid_by_epoch = {}               # epoch -> validation loss

    with open(log_path) as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            try:
                rec = json.loads(line)
            except json.JSONDecodeError:
                continue  # skip any non-JSON lines

            mode = rec.get("mode")
            epoch = rec.get("epoch")
            loss = rec.get("loss")
            # Skip malformed records and the initial eval (epoch is null)
            if loss is None or epoch is None:
                continue

            if mode == "opt":
                opt_by_epoch[epoch].append(loss)
            elif mode == "eval":
                valid_by_epoch[epoch] = loss

    train_epochs = sorted(opt_by_epoch)
    train_loss = [sum(opt_by_epoch[e]) / len(opt_by_epoch[e]) for e in train_epochs]

    valid_epochs = sorted(valid_by_epoch)
    valid_loss = [valid_by_epoch[e] for e in valid_epochs]

    return train_epochs, train_loss, valid_epochs, valid_loss


def main(log_path, out_path="training_curves.png"):
    result = parse_loss_log(log_path)

    if not result or not result[0]:
        print("No log data found — run a training job first.")
        return

    train_epochs, train_loss, valid_epochs, valid_loss = result

    fig, ax = plt.subplots(figsize=(8, 4))
    ax.semilogy(train_epochs, train_loss, "-o", label="Train loss (epoch mean)")
    ax.semilogy(valid_epochs, valid_loss, "-s", label="Validation loss")
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Loss (log scale)")
    ax.set_title("X-MACE Training Curves")
    ax.legend()
    plt.tight_layout()
    plt.savefig(out_path, dpi=150)
    print(f"Saved plot to {out_path}")
    plt.show()


if __name__ == "__main__":
    # Path to the X-MACE results log (JSON-lines). Default name pattern is
    # results/<name>_run-<seed>.json  e.g. results/energies_forces_run-100.json
    # Override on the command line:  python plot_loss.py results/my_run.json
    default_log = "results/oscillator_stengths_run-100_train.txt"
    log_path = sys.argv[1] if len(sys.argv) > 1 else default_log
    main(log_path)
```

## 4. Loading the Model and Predicting Oscillator Strengths

After training, a model file named `oscillator_strengths.model` is saved to the working directory (and checkpoints to `checkpoints/`). Load it with `MACECalculator` exactly as you would an energy model — the only difference is that `calc.results["energy"]` now contains oscillator strengths instead of electronic state energies.

```python
import torch
import ase.io
import numpy as np
import sys
sys.path.insert(0, ".")
from mace import MACECalculator

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

MODEL_PATH = "oscillator_strengths.model"   # <-- update if needed
N_OSC      = 3                               # must match --n_energies used in training

calc = MACECalculator(
    model_paths=MODEL_PATH,
    device=device,
    n_energies=N_OSC,         # the model predicts N_OSC scalar values per frame
    default_dtype="float32",
    model_type="MACE",
)
print("Calculator loaded successfully.")
```

### 4a. Predict for a single geometry

```python
# Load a single frame
atoms = ase.io.read("oscillator_strengths.xyz", index=0)

# Attach the calculator and trigger a prediction
atoms.calc = calc
atoms.get_potential_energy()   # standard ASE call; internally runs the model

# Predictions are in calc.results['energy'] — NOT in atoms.info
pred_osc = np.array(calc.results["energy"]).flatten()   # shape (N_OSC,)

# Reference values are stored in REF_energy in the XYZ file
ref_osc = np.array(atoms.info['REF_energy']).flatten()

print("=== Predicted vs Reference Oscillator Strengths ===")
print(f"{'Transition':>12}  {'Predicted':>12}  {'Reference':>12}  {'Error':>10}")
print("-" * 52)
transitions = [f"S0\u2192S{i+1}" for i in range(N_OSC)]
for label, pred, ref in zip(transitions, pred_osc, ref_osc):
    print(f"{label:>12}  {pred:>12.6f}  {ref:>12.6f}  {abs(pred-ref):>10.6f}")
```

### Understanding the output

| `calc.results` key | Shape | Meaning |
|---|---|---|
| `energy` | `(N_OSC,)` | Predicted oscillator strength for each transition |
| `free_energy` | `(N_OSC,)` | Same as `energy` (alias used by ASE internals) |

Note that `forces` will also be present in `calc.results` but is **physically meaningless** for oscillator strengths — ignore it. The model was trained with `forces_weight=0.0`, so those values are not trained and should not be used.

## 5. Running Predictions Over a Dataset

The workflow below iterates over all frames, collects predicted and reference oscillator strengths, and computes per-transition MAE.

```python
# Load the dataset (or a subset for speed)
frames = ase.io.read("oscillator_strengths.xyz", index=":")

pred_osc_all = []   # (n_frames, N_OSC)  — from calc.results['energy']
ref_osc_all  = []   # (n_frames, N_OSC)  — from atoms.info['REF_energy']

print(f"Running predictions on {len(frames)} frames ...")
for i, frame in enumerate(frames):
    frame.calc = calc
    frame.get_potential_energy()   # triggers calculate()

    # Predictions come from calc.results, not atoms.info
    pred_osc_all.append(np.array(calc.results["energy"]).flatten())

    # Reference values live in REF_energy — atoms.info['energy'] does not exist
    ref = frame.info.get("REF_energy", None)
    if ref is not None:
        ref_osc_all.append(np.array(ref).flatten())

    if (i + 1) % 50 == 0:
        print(f"  Processed {i + 1}/{len(frames)} frames")

pred_osc_all = np.array(pred_osc_all)   # (n_frames, N_OSC)
ref_osc_all  = np.array(ref_osc_all)    # (n_frames, N_OSC)

print(f"\nPredictions shape : {pred_osc_all.shape}")
print(f"References shape  : {ref_osc_all.shape}")
```

### 5a. Per-transition MAE

```python
if len(ref_osc_all) > 0:
    mae_per_transition = np.abs(pred_osc_all - ref_osc_all).mean(axis=0)   # (N_OSC,)
    mae_overall        = mae_per_transition.mean()

    print("Per-transition MAE:")
    for i, mae in enumerate(mae_per_transition):
        print(f"  S0→S{i+1}: {mae:.5f}")
    print(f"\nOverall MAE: {mae_overall:.5f}")
else:
    print("No reference data found — skipping MAE calculation.")
```

### 5b. Correlation plots — predicted vs reference

```python
import matplotlib.pyplot as plt

if len(ref_osc_all) > 0:
    fig, axes = plt.subplots(1, N_OSC, figsize=(4 * N_OSC, 4), sharey=False)
    if N_OSC == 1:
        axes = [axes]

    for i, ax in enumerate(axes):
        x = ref_osc_all[:, i]
        y = pred_osc_all[:, i]
        lims = [min(x.min(), y.min()), max(x.max(), y.max())]

        ax.scatter(x, y, s=8, alpha=0.5, label="frames")
        ax.plot(lims, lims, "r--", linewidth=1, label="y = x")
        ax.set_xlabel(f"Reference f(S0→S{i+1})")
        ax.set_ylabel(f"Predicted f(S0→S{i+1})")
        ax.set_title(f"S0→S{i+1}  MAE={mae_per_transition[i]:.4f}")
        ax.legend(fontsize=7)

    fig.suptitle("Oscillator Strength: Predicted vs Reference", fontsize=12)
    plt.tight_layout()
    plt.show()
else:
    print("No reference data found — skipping correlation plots.")
```

## 6. Tips and Troubleshooting

**Normalisation.** Oscillator strengths can span several orders of magnitude across frames (e.g. near-zero at dark states vs. large at bright states). If training is unstable, consider normalising each transition independently before writing the XYZ, and denormalising after prediction.

**Number of outputs.** `--n_energies` must exactly match `N_OSC`. If you have 6 unique pairwise oscillator strengths (all 4-state pairs), set `--n_energies=6`.

**Joint training.** You can train oscillator strengths alongside conventional multi-state energies by concatenating them into a single longer vector, e.g. `[E_S0, E_S1, E_S2, E_S3, f_01, f_02, f_03]` and setting `--n_energies=7`. In that case you do want `--forces_weight > 0`, but forces will only be defined for the energy components — this requires care when interpreting `calc.results["forces"]`.

**Model choice.** `AutoencoderExcitedMACE` is recommended for better accuracy. `ExcitedMACE` is faster and sufficient for quick tests.

**`forces` in results.** When predicting with a model trained on oscillator strengths, `calc.results["forces"]` will be present but untrained and physically meaningless — always ignore it for oscillator strength models.

---

**Quick-reference flag summary for oscillator strength training:**

| Flag | Value | Reason |
|---|---|---|
| `--n_energies` | `N_OSC` | One output per oscillator strength value |
| `--energy_weight` | `100.0` | The only active loss term |
| `--forces_weight` | `0.0` | No forces for oscillator strengths |
| `--nacs_weight` | `0.0` | Not applicable |
| `--socs_weight` | `0.0` | Not applicable |
| `--model` | `AutoencoderExcitedMACE` | Best accuracy for scalar properties |
